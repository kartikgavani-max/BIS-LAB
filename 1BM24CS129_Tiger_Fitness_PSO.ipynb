```python
import numpy as np

def tiger_habitat_fitness(position):
    prey_density = position[0]
    forest_cover = position[1]
    water_distance = position[2]
    road_distance = position[3]

    if prey_density < 5 or forest_cover < 20 or water_distance > 15:
        return -99999

    prey_score = -((prey_density - 25) ** 2)
    forest_score = -((forest_cover - 75) ** 2)
    water_score = -((water_distance - 1) ** 2)

    if road_distance < 3:
        human_penalty = -1000
    elif road_distance < 8:
        human_penalty = -250
    else:
        human_penalty = 0

    return prey_score + forest_score + water_score + human_penalty

num_particles = 30
num_dimensions = 4
max_iterations = 60
inertia_weight = 0.6
cognitive_coeff = 1.7
social_coeff = 1.7

lower_bounds = np.array([5.0, 0.0, 0.0, 0.0])
upper_bounds = np.array([50.0, 100.0, 15.0, 20.0])

particle_positions = np.random.uniform(lower_bounds, upper_bounds, (num_particles, num_dimensions))
particle_velocities = np.random.uniform(-1, 1, (num_particles, num_dimensions))

personal_best_positions = np.copy(particle_positions)
personal_best_fitness = np.array([tiger_habitat_fitness(p) for p in particle_positions])

global_best_index = np.argmax(personal_best_fitness)
global_best_position = np.copy(personal_best_positions[global_best_index])
global_best_fitness = personal_best_fitness[global_best_index]

for iteration in range(max_iterations):
    for i in range(num_particles):
        current_fitness = tiger_habitat_fitness(particle_positions[i])

        if current_fitness > personal_best_fitness[i]:
            personal_best_fitness[i] = current_fitness
            personal_best_positions[i] = np.copy(particle_positions[i])

            if current_fitness > global_best_fitness:
                global_best_fitness = current_fitness
                global_best_position = np.copy(particle_positions[i])

        r1 = np.random.rand(num_dimensions)
        r2 = np.random.rand(num_dimensions)

        cognitive_velocity = cognitive_coeff * r1 * (personal_best_positions[i] - particle_positions[i])
        social_velocity = social_coeff * r2 * (global_best_position - particle_positions[i])

        particle_velocities[i] = (inertia_weight * particle_velocities[i]) + cognitive_velocity + social_velocity
        particle_positions[i] += particle_velocities[i]

        particle_positions[i] = np.clip(particle_positions[i], lower_bounds, upper_bounds)

print("=== Tiger Habitat Optimization Complete ===")
print(f"Optimal Prey Density:       {global_best_position[0]:.2f} animals/km²")
print(f"Optimal Forest Cover:       {global_best_position[1]:.2f}%")
print(f"Optimal Distance to Water:  {global_best_position[2]:.2f} km")
print(f"Optimal Distance to Roads:  {global_best_position[3]:.2f} km")
print(f"Final Habitat Fitness Score: {global_best_fitness:.4f}")
```
