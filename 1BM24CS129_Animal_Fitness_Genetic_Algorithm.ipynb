```python

import random

def fitness(habitat):
    food, water, shelter = habitat
    return food + water + shelter

POPULATION_SIZE = 6
GENERATIONS = 20
MUTATION_RATE = 0.1

population = [
    [random.randint(0, 10) for _ in range(3)]
    for _ in range(POPULATION_SIZE)
]

for generation in range(GENERATIONS):

    population.sort(key=fitness, reverse=True)

    new_population = [population[0][:]]

    while len(new_population) < POPULATION_SIZE:
        parent1, parent2 = random.choices(population[:3], k=2)

        point = random.randint(1, 2)
        child = parent1[:point] + parent2[point:]

        for i in range(3):
            if random.random() < MUTATION_RATE:
                child[i] = random.randint(0, 10)

        new_population.append(child)

    population = new_population

    best = max(population, key=fitness)
    print(
        f"Generation {generation + 1}: "
        f"Habitat = {best}, Fitness = {fitness(best)}"
    )

best = max(population, key=fitness)

print("\n--- Final Result ---")
print("Best Tiger Habitat:", best)
print("Food Score:", best[0])
print("Water Score:", best[1])
print("Shelter Score:", best[2])
print("Maximum Fitness:", fitness(best), "/ 30")
```
